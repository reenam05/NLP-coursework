NLP Approach
The set of rules used is:
- if the email/sentence contains key words which related to spam that are in the list "spam_words", the likelihood of it being spam increases

- if the email/sentence contains links which related to spam that are in the list "spam_links", the likelihood of it being spam increases

- if the email/sentence contains punctuation which related to spam that are in the list "spam_punct", the likelihood of it being spam increases

- if the email/sentence contains key words, links and punctuation which are NOT related to spam that are in the list "not_spam", the likelihood of it being spam decreases

- after each checks, if the the likelihood of the sentence being spam surpasses a threshold, the sentence is predicted as spam.

```python
def spam_detection(s):
    #spam count
    spam = 0

    #SCOPE OF THE RULES
    #these are key words which are very common in spam messages
    #spam sentences usually contain lots of capital letters and certain punctuations to grab the users attention,
    #hyperlinks are also used to scam users, either by obtaining their information or persuading them to purchase an item
    #these attributes are saved in a list and will be compared to the sentences to predict spam.
    spam_words = ["win","winner","congratulations","free","lucky","won","hurry","sale","ends","limited","virus","bank","send", "prizes",
                  "WIN","WINNER","CONGRATULATIONs","FREE","LUCKY","WON","HURRY","SALE","ENDS","LIMITED","VIRUS","BANK","SEND"]
    spam_links = ["www.",".com","https",".co.uk"]
    spam_punct = ["!!!","!!","???","?!!"]

    #it is common for companies the user has signed up to, or other non spam content to include hyperlinks, punctuation and certain key words.
    #the algorithm also considers whether the sentences is not spam
    not_spam = ["check", "www.",".com","https",".co.uk","!",".","?","reminder","I", "my"]

    # the first checks is for key spam words as this is one of the biggest indicators for spam
    for word in spam_words:
        if word in s:
            spam = spam + 1
    
    #next check is for spam links.
    for word in spam_links:
        if word in s:
            #If key spam words have already been detected and spam links are detected, it makes the email even more likely to be spam.
            if spam > 2:
                #to highlight the greater likelihood, we increment spam by 2
                spam = spam + 2
            else:
                spam = spam + 1

    #next check is for common punctuation used for spam
    for word in spam_punct:
        if word in s:
            # if the email has spam key words or links, it is more likely for the email to be spam if it also has certain punctuation.
            if spam > 3:
                #to highlight the greater likelihood, we increment spam by 3
                spam = spam + 3
            else:
                spam = spam + 1

    #final check is to detect for non spam key words, punctuation and links. 
    for word in not_spam:
        if word in s:
            spam = spam - 2

    if spam > 2:
        print("This sentence is likely spam:",s,spam)
    else:
        print("Not likely to be spam:",s,spam)

#WHAT CONSTITUTE AS SPAM:
# bad deals and discounts which the user has not willingly signed up
# emails with malicious links and attachment
# promotional messages from unknown sources/companies
# malicious emails asking user's for their bank details/ money for personal use

#general testcases
b12 = "You could be our lucky winner if you sign up today and call us to gain free prizes SPAM"#SPAM
b13 = "can I set a meeting for this friday we need to work on our deal" #NOT SPAM
b14 = "Sale ends this Wednesday don't miss out and send us your bank details SPAM" #SPAM
b15 = "here is some information you may like to read www.historicalevents.com"#NOT SPAM

##Specific edgecases
##company promoting a good deal - missclassification to spam
b1 = "Limited offer! Buy now and get 10% off"#NOT SPAM
b2 = "Because you signed up, here is a 5% discount on your next meal!"#NOT SPAM

##informational messages with links - - missclassification to spam
b3 = "Many students are known to like travelling from the data www.studentdata.com"#NOT SPAM
b4 = "As you requested information on banks opening times, here is the link www.banks.com/time"#NOT SPAM

##using punctuation like ?, ! and exciting words - missclassification to spam
b5 = "It’s my birthday I am so excited, going to go shopping!" #NOT SPAM
b6 = "Honestly I am so surprised I got this amazing job offer, I am so happy!"#NOT SPAM

#short sentences - harder for algorithm to detect spam
b7 = "HURRY LUCKY WINNER!!!" #SPAM
b8 = "limited sale ends today?!!" #SPAM

#FALSE POSITIVE AND FALSE NEGATIVES
# spammers intentionally write thier emails to pass the spam algorithms. 
# they do this by presenting themselves as a legitimate company by using "non spam" language, punctuation and links.
# false positives occur when a non spam email gets treated as spam
# this occurs when users unintentionally write emails which pass the checks of spam by using spam key words, punctuation and links 
b9 = "Buy one get three free today limited sale ends today. SPAM" #SPAM -- False negative
b10 = "Today you could be the lucky winner, sign up to www.winmoney.com SPAM"#SPAM -- False negative
b11 = "Today I went shopping, then to a cafe, and then to a museum, it was a lot of fun" #NOT SPAM - false positive

spam_detection(b1)
spam_detection(b2) 
spam_detection(b3)
spam_detection(b4)
spam_detection(b5)
spam_detection(b6)
spam_detection(b7)
spam_detection(b8)
spam_detection(b9)
spam_detection(b10)
spam_detection(b11)
spam_detection(b12)
spam_detection(b13)
spam_detection(b14)
spam_detection(b15)
```

- this method uses Hobbs algorithm which is rule based and focuses of comparing NP by their gender, persons and number.
- It also presumes that the antecedent comes before the anaphor and that they two would be close together in the sentence.

NLP Approach
- For this task I used the CRF model to identify and classify entities.
- Conditional Random Fields are used for sequence labelling tasks such as NER.
- The model is interpretable and faster compared to other models like neural networks.
- It also performs relatively well in evaluations.
- CRF predicts the entire sequence of labels simultaneously rather than individually for each token.
- It also avoids the label bias problem.

- The dataset used was the CoNLL-2002 which has been preprocessed into samples with each token labelled. 
- The tokens are labelled using the BIO format.
- The model uses features from each token which was manually obtained as shown below.
- During training the model chooses the most likely label sequence through combining feature weights and transition dependencies across the sentence
- Which leads to the model predicting the most optimal/ most likely label sequence for the testing data

```python
import sklearn_crfsuite
from sklearn_crfsuite import metrics
import nltk
import numpy as np
from sklearn.model_selection import train_test_split

nltk.download('conll2002')
conll = nltk.corpus.conll2002.iob_sents()

def features(sentence, i):
    # set of features we want to know from each token
    word_feature = {
        'sentence': sentence,
        'first word': sentence == sentence[0],
        'last word': sentence == sentence[len(sentence)-1],
        'all caps':  sentence == np.char.upper(sentence),
        'all lowercase':  sentence == np.char.lower(sentence),
        'suffix last letter': np.array([s[-1] for s in sentence]),
        'suffix last two letters': np.array([s[-2:-1] for s in sentence]),
        'suffix last three letters': np.array([s[-3:-1] for s in sentence]),
        'word is a title': sentence == np.char.istitle(sentence),
    }
    #takens in a sentences as a vector of tokens and applies each rule to each token
    #this obtains the features for each token
    arr = []
    for i in range(len(word_feature['sentence'])):
        dict =  {key: value[i] for key, value in word_feature.items()}
        arr.append(dict)
    return arr

sentence_info = []
tags = []

#assumptions
# model also allows us to be flexible when choosing what features we want to apply for each sample of data
# there are no independence assumptions because it uses overlapping features 
for key, sentence in enumerate(conll):
    words = np.array(sentence)
    word_vec = words[:, 0]
    ner_vec = words[:, 2]
    index = np.arange(len(word_vec))  
    sentence_feature = features(word_vec,index)
    sentence_info.append(sentence_feature)
    tags.append(ner_vec)

# Split dataset
# the training to testing set is a 80:20 ratio which we assume will be suitable for the model to predict effectively.
X_train, X_test, y_train, y_test = train_test_split(sentence_info, tags,test_size=0.2,random_state=42)

## ASSUMPTIONS about entity categories
## we are training the model on 5 categories, PER, ORG, LOC, MISC, O
## the model assumes that for each sample, the features are extracted appropriately
#model predicts the entire sequence of labels together instead of outputting each tokens independently 
crf = sklearn_crfsuite.CRF(algorithm='lbfgs', c1=0.1, c2=0.1, max_iterations=100, all_possible_transitions=True)
crf.fit(X_train, y_train)

labels = list(crf.classes_)  
labels.remove('O')

#predictions
y_pred = crf.predict(X_test)

#evaluations
report = metrics.flat_classification_report(y_test, y_pred, labels=labels, digits=3,zero_division=0  )
print(report)

print(metrics.flat_accuracy_score(y_test, y_pred))
```

- Perhaps a longer list of features may increase the scores.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

df = pd.read_csv("C:/Users/20men/vs/practising/archive (1)/olid-training-v1.0.tsv",sep="\t")

X = df["tweet"]
y = df["subtask_a"]

#TF-IDF works well with the SVM model as it converts samples of data into a numerical vector
# the vectors are high dimensional.
vectorizer = TfidfVectorizer(stop_words='english', max_df=0.7)
X_vector = vectorizer.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_vector, y, test_size=0.2, random_state=42)

# we use SVM because it is a intepretable and quick model compared to other models
# it is well suited for a task such as binary classification for offensive language detection
classifer = SVC(kernel='linear')
classifer.fit(X_train, y_train)

# Predictions
#Assumptions: the dataset assumes that insults, threats, derogatory comments, profanity is offensive language
y_pred = classifer.predict(X_test)

# Evaluate the performance on precision recall and f1 - scores
report = classification_report(y_test, y_pred)

# Evaluations
accuracy = accuracy_score(y_test, y_pred)
print(report)

print(f'Accuracy: {accuracy:.4f}')
```

Results and analysis
- From the OLID dataset we used the data of text and the labels regarding if they were offensive or not
- The accuracy shows that the model is learning and generalising patterns well.
- It can also be implied that the model is not overfitting.
- Some common missclassification could be due to text which have hidden meanings for the model. For an example, it may not classify sarcasm properly. 
- The model performed well in precision, recall and f1 scores for predicting not offensive text.
- It performed adequately for recall and f1-scores for predicting offensive texts.
- Perhaps, to improve performance, the dataset could be manually transformed to remove emojis, URL etc
- Furthermore, instead of using N-grams for short phrases may be beneficial for accuracy
- Or using more advanced models such as BERT as it can learn more difficult patterns and context, but then the comprise of run time would have to be considered.
